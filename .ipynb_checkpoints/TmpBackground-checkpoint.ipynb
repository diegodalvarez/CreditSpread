```python
import os
import pandas as pd
import matplotlib.pyplot as plt
```

```python
path = r"A:\2026BlpAdHocData\Combined\FixedIncomeIndex\BloombergOAS.parquet"
df_raw = (pd
    .read_parquet(path = path, engine = "pyarrow"))
```

```python
df_oas_ticker = (df_raw
    [["security"]]
    .groupby("security")
    .head(1)
    .assign(str_ending = lambda x: x.security.str.split(" ").str[0].str[-3:])
    .loc[lambda x: x.str_ending == "OAS"])
```

```python
oas_path = r"A:\2026BlpAdHocData\Combined\FixedIncomeIndex\BloombergOAS.parquet"
df_oas   = (pd
    .read_parquet(path = oas_path, engine = "pyarrow")
    .assign(
        str_ending = lambda x: x.security.str.split(" ").str[0].str[-3:],
        truu_ticker = lambda x: x.security.str.split(" ").str[0].str[:-3] + "TRUU Index")
    .loc[lambda x: x.str_ending == "OAS"])
```

```python
tr_path = r"A:\2026BlpAdHocData\Combined\FixedIncomeIndex\BloombergTR.parquet"
df_tr   = (pd
    .read_parquet(path = tr_path, engine = "pyarrow")
    .rename(columns = {"PX_LAST": "truu_px"}))
```

```python
sprd_dur_path = r"A:\2026BlpAdHocData\Combined\FixedIncomeIndex\INDEX_BLENDED_SPREAD_DUR.parquet"
df_sprd_dur   = (pd
    .read_parquet(path = sprd_dur_path, engine = "pyarrow")
    .rename(columns = {"INDEX_BLENDED_SPREAD_DUR": "sprd_dur"}))
```

```python
df_combined = (df_tr
    .merge(right = df_sprd_dur, how = "inner", on = ["date", "security"]))
```
