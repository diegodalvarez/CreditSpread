# Duration Times Spread Strategies

```python
note_path = os.getcwd()
repo_path = os.path.abspath(os.path.join(note_path, ".."))
data_path = os.path.join(repo_path, "data")
```

```python
ticker_path   = os.path.join(data_path, "CreditTickers.xlsx")
plot_renamer_ = (pd
    .read_excel(io = ticker_path)
    .assign(ticker = lambda x: x.ticker.str.split(" ").str[0])
    .set_index("ticker")
    .plot_name
    .to_dict())

renamer = {**plot_renamer_}
```

# Strategy 1: High Yield Risk Premia Compensation

High yield securities are quoted in prices as opposed to the investment grade bonds which are quoted in OAS because high yield securities tend to have are more equity-like component. *Duration-time-spread* (DTS) offers are new risk measurement for credit modelling that is universal across bond ratings, and has a much better accuracy in modelling risk. Therefore we can develop a strategy that profits on getting componesated for DTS. The idea is that we can measure spread risk in terms of OAS and bond risk in terms of DTS. Z-scoring these risk measures and taking their difference yield us an indicator for getting compensated for the risk. For example OAS may widen implying more risk but if DTS doesn't widen as much then we can deploy capital getting compensated for that risk and vice versa.

```python
tickers = ['I00185US','I09036US','LF98TRUU','LHVLTRUU']
```

```python
path   = os.path.join(data_path, "SlicedCreditData.parquet")
df_raw = (pd
    .read_parquet(path = path, engine = "pyarrow")
    .assign(security = lambda x: x.security.str.split(" ").str[0])
    .loc[lambda x: x.security.isin(tickers)])
```

```python
start_date = (df_raw
    .pivot(index = "date", columns = "security", values = "oas")
    .reset_index()
    [["date"]]
    .assign(
        lag_date    = lambda x: x.date.shift(),
        spread_date = lambda x: (x.date - x.lag_date).dt.days)
    .loc[lambda x: x.spread_date <= 4]
    .date
    .iloc[0])
```

```python
df_sliced = (df_raw
    .loc[lambda x: x.date >= start_date])
```

```python
window = 100

df_signal = (df_sliced
    [["date", "security", "oas", "sprd_dur"]]
    .set_index("date")
    .assign(dur_oas = lambda x: x.sprd_dur * x.oas)
    .drop(columns = ["sprd_dur"])
    .groupby("security")
    .apply(lambda x: (x - x.ewm(span = window, adjust = False).mean()) / x.ewm(span = window, adjust = False).std())
    .assign(signal = lambda x: x.oas - x.dur_oas)
    .reset_index()
    .pivot(index = "date", columns = "security", values = "signal")
    .shift()
    .reset_index()
    .melt(id_vars = "date", value_name = "lag_signal")
    .dropna())
```

```python
df_signal_rtn = (df_sliced
    .pivot(index = "date", columns = "security", values = "truu")
    .pct_change()
    .reset_index()
    .melt(id_vars = "date", value_name = "px_rtn")
    .merge(right = df_signal, how = "inner", on = ["date", "security"])
    .assign(signal_rtn = lambda x: np.sign(x.lag_signal) * x.px_rtn))
```

```python
df_vol_rtn = (df_signal_rtn
    .pivot(index = "date", columns = "security", values = "signal_rtn")
    .apply(lambda x: x * (0.1 / (x.ewm(span = 100, adjust = False).std() * np.sqrt(252))))
    .reset_index()
    .melt(id_vars = "date", value_name = "vol_rtn")
    .dropna())
```

```python
(df_vol_rtn
    .rename(columns = {"security": ""})
    .pivot(index = "date", columns = "", values = "vol_rtn")
    .apply(lambda x: np.cumprod(1 + x))
    .rename(columns = renamer)
    .plot(
        figsize = (8,5),
        logy    = True,
        ylabel  = "Cuml. Ret. Log-Scaled",
        title   = "Cumulative Returns of Trading Residual DTS Risk Compenstaion for HY Indices\n10% Annualized Volatility (Hedged Contemporaneously)"))

plt.show()
```

```python
(df_vol_rtn
    .drop(columns = ["date"])
    .groupby("security")
    .agg(lambda x: x.mean() / x.std() * np.sqrt(252))
    .T
    .assign(Avg = lambda x: x.mean(axis = 1)))
```

Now with a collection of high yield ETFs the same approach will be taken. Getting the OAS and spread duration is difficult and thus the data from the indices will be used. Matching for index and each ETF's benchmark won't be used therefore there will be $m \times n$ number of strategies. Start by naively trading which is taking the position proportional to the signal of the adjusted returns with a 10% target allocation

```python
etf_path   = os.path.join(data_path, "HYETFs.parquet")
df_etf_rtn = (pd
    .read_parquet(path = etf_path, engine = "pyarrow")
    .loc[lambda x: x.Price == "Adj Close"]
    .pivot(index = "Date", columns = "Ticker", values = "value")
    .pct_change()
    .reset_index()
    .melt(id_vars = "Date", value_name = "px_rtn")
    .dropna()
    .rename(columns = {"Date": "date"}))
```

```python
df_etf_signal_rtn = (df_etf_rtn
    .merge(right = df_signal, how = "inner", on = ["date"])
    .assign(
        name       = lambda x: x.Ticker + " " + x.security,
        signal_rtn = lambda x: np.sign(x.lag_signal) * x.px_rtn))
```

```python
df_vol_rtn = (df_etf_signal_rtn
    .pivot(index = "date", columns = "name", values = "signal_rtn")
    .apply(lambda x: x * (0.1 / (x.ewm(span = 100, adjust = False).std() * np.sqrt(252)))))
```

The performance is ok, but doesn't show strong returns across the board on the other hand the strategy is quite naive and thus the lack of overperformance shouldn't be taken lightly. 

```python
(df_vol_rtn
    .apply(lambda x: np.cumprod(1 + x))
    .plot(
        figsize = (8,5),
        logy    = True, 
        legend  = False,
        alpha   = 0.7, 
        ylabel  = "Cuml. Ret. Log-Scaled", 
        title   = "Cumulative Returns of Trading Residual DTS Risk Compenstaion for HY ETFs\n10% Annualized Volatility (Hedged Contemporaneously)"))

plt.show()
```

There appears to be a performance gap between ETF provider and index choice from the table below. 

```python
(df_vol_rtn
    .agg(lambda x: x.mean() / x.std() * np.sqrt(252))
    .to_frame(name = "sharpe")
    .reset_index()
    .assign(
        etf      = lambda x: x.name.str.split(" ").str[0],
        security = lambda x: x.name.str.split(" ").str[1])
    .pivot(index = "security", columns = "etf", values = "sharpe"))
```

```python
df_vol_longer = (df_vol_rtn
    .reset_index()
    .melt(id_vars = "date")
    .assign(
        ticker = lambda x: x.name.str.split(" ").str[0],
        bench  = lambda x: x.name.str.split(" ").str[1]))
```

```python
df_control_etf = (df_vol_longer
    [["date", "ticker", "value"]]
    .groupby(["date", "ticker"])
    .agg("mean")
    .reset_index()
    .rename(columns = {"ticker": ""}))

df_control_bench = (df_vol_longer
    [["date", "bench", "value"]]
    .groupby(["date", "bench"])
    .agg("mean")
    .reset_index()
    .rename(columns = {"bench": ""}))
```

```python
plot_dict = {
    "ETF"      : df_control_etf,
    "Benchmark": df_control_bench}

fig, axes = plt.subplots(ncols = len(plot_dict.keys()), figsize = (20,6))
for key, ax in zip(plot_dict.keys(), axes.flatten()):

    (plot_dict[key]
        .pivot(index = "date", columns = "", values = "value")
        .assign(Avg = lambda x: x.mean(axis = 1))
        .apply(lambda x: np.cumprod(1 + x))
        .plot(
            logy   = True, 
            ax     = ax,
            title  = key,
            ylabel = "Cuml. Ret. Log-Scaled"))
```
